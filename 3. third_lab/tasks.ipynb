```python
import numpy as np
import pandas as pd
```

```python
population_dict = {'Moscow': 13364800,
                   'Novosibirsk': 1637266,
                   'Vladivostok': 610628,
                   'Omsk': 1101367,
                   'Tomsk': 544566,
                   'Irkutsk': 605708}

population = pd.Series(population_dict)
area_dict = {'Moscow': 2561,
             'Novosibirsk': 502.7,
             'Vladivostok': 561.5,
             'Omsk': 583.3,
             'Tomsk': 295.1,
             'Irkutsk': 280}
area = pd.Series(area_dict)

cities = pd.DataFrame({'population': population,
                       'area': area})
cities['density'] = cities['population'] / cities['area']
cities
```
